```python
import numpy as np
import matplotlib.pyplot as plt
import os
import glob
import random

# Ruta relativa a la carpeta on tens les dades (així funciona tant en Windows com en Linux/servidor)
data_dir = "./data_processed"

images_paths = sorted(glob.glob(os.path.join(data_dir, "images", "*.npy")))
masks_paths = sorted(glob.glob(os.path.join(data_dir, "masks", "*.npy")))

print(f"S'han trobat {len(images_paths)} talls útils processats.")

# Triem una imatge aleatòria del dataset
idx = random.randint(0, len(images_paths) - 1)
img = np.load(images_paths[idx])
mask = np.load(masks_paths[idx])

# Dibuixem els 4 canals de la MRI i la màscara
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
titles = ['T1', 'T1ce', 'T2', 'FLAIR', 'Màscara (Tumor)']

# Mostrar els canals de la ressonància
for i in range(4):
    axes[i].imshow(img[i], cmap='gray')
    axes[i].set_title(titles[i])
    axes[i].axis('off')

# Mostrar la màscara sobre fons de colors per ressaltar
axes[4].imshow(mask[0], cmap='nipy_spectral')
axes[4].set_title(titles[4])
axes[4].axis('off')

plt.suptitle(f"Visualitzant arxiu: {os.path.basename(images_paths[idx])}", fontsize=14)
plt.tight_layout()
plt.show()
```
